```python
import pandas as pd
import numpy as np
```

```python
s10 = pd.Series(np.random.randint(50, 151, 20))
print("Series:", s10.tolist())
print("Сумма:", s10.sum())
print("Среднее:", s10.mean())
print("Минимум:", s10.min())
print("Максимум:", s10.max())
print("Стандартное отклонение:", s10.std())
```
